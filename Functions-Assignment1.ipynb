```python
def OddEven():
    num=int(input("Enter a number:"))
    if(num%2==0):
        print(num ,"is Even number")
    else:
        print(num ,"is ODD number")
```

```python
OddEven()
```

```python
def percentage():
    list=[98,87,95,95,93]
    total = 0
    for i in range(len(list)):
        print("Subject",i,"=",list[i])
        total += list[i]
    print("Total=",total)
    percentage=total/len(list)
    print("Percentage =", percentage, "%")
    
```

```python
percentage()
```
